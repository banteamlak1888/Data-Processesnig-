```python
import re
import csv
from google.colab import files

# Define the tokenization pattern for Ethiopic characters + punctuation
pattern = r"[\u1200-\u137F]+|[።፣፤፥፦፧፨]"

# Open your dataset file (make sure 'dataset.txt' is in the same folder)
with open("corpus2.txt", "r", encoding="utf-8") as file:
    sentences = file.readlines()

# Create CSV file
with open("tokenized_output.csv", "w", encoding="utf-8", newline="") as csv_file:
    writer = csv.writer(csv_file)
    writer.writerow(["Sentence_ID", "Tokens"])  # header

    # Tokenize each sentence
    for i, sentence in enumerate(sentences, 1):
        sentence = sentence.strip()  # remove newlines/spaces
        if not sentence:
            continue  # skip empty lines

        tokens = re.findall(pattern, sentence)
        print(tokens)
        writer.writerow([i, " ".join(tokens)])  # write tokens to CSV

# Download the file in Colab
files.download("tokenized_output.csv")
print("✅ Tokenized CSV file ready for download: tokenized_output.csv")
```
