```python
import base64
import pandas as pd
from IPython.display import HTML

def transform_to_20_classes_and_download(input_excel_file, output_excel_file):
    """
    Collapses 39 BIO tag boundaries into 20 clean entity categories
    and automatically builds an inline browser download hyperlink.
    """
    # 1. Load the original dataset layout
    df = pd.read_excel(input_excel_file)

    # Cast to safe string type to prevent floating-point disruptions
    df['NER_Tag'] = df['NER_Tag'].astype(str).str.strip()

    # 2. Define the exact categorical mapping function
    def strip_bio_prefix(tag):
        tag_str = str(tag).strip()

        # If outside or clean, return as is
        if tag_str == 'O' or tag_str == 'nan':
            return 'O'

        # Strip BIO boundaries (e.g., 'B-GOD' or 'I-GOD' becomes 'GOD')
        if tag_str.startswith('B-') or tag_str.startswith('I-'):
            return tag_str[2:]

        return tag_str

    # 3. Create the clean 20-class feature vector column
    # We create a new column to leave your original data columns perfectly safe
    df['NER_Class_20'] = df['NER_Tag'].apply(strip_bio_prefix)

    # 4. Save the modified dataset workbook matrix locally
    df.to_excel(output_excel_file, index=False)

    print("🧱 ===============================================================")
    print(f"📊   STAGE 1 RE-MAPPING COMPLETE: UNIQUE NER TARGET CLASSES = {df['NER_Class_20'].nunique()}")
    print("===============================================================")
    print(df['NER_Class_20'].value_counts().to_string())
    print("===============================================================")
    print("🔒 All other data paths remained perfectly preserved.")

    # 5. GENERATE INSTANT BROWSER DOWNLOAD LINK
    with open(output_excel_file, 'rb') as f:
        excel_bytes = f.read()

    b64_encoded_excel = base64.b64encode(excel_bytes).decode()
    download_href = (
        f'<a href="data:application/vnd.openxmlformats-officedocument.spreadsheetml.sheet;base64,{b64_encoded_excel}" '
        f'download="{output_excel_file}" style="font-weight: bold; color: #E65100; font-size: 16px;">'
        f'👉 Click Here to Download "{output_excel_file}" Directly To Your Machine 👈</a>'
    )

    return HTML(download_href)

# --- Runtime Invocation Configuration ---
# Match these filenames to your workspace path
source_sheet = "0MLT__Data.xlsx"
target_sheet = "0MLT__Data_20NER.xlsx"

# Run mapping pipeline and render the download interface
transform_to_20_classes_and_download(source_sheet, target_sheet)
```

```python
import base64
import numpy as np
import pandas as pd
from IPython.display import HTML

def calculate_ner_weights_and_download(input_excel_file, output_txt_file):
    """
    Calculates inverse-frequency class weights for the 39-class BIO column
    to eliminate model overfitting without altering the target labels.
    """
    # 1. Load the pristine dataset sheet
    df = pd.read_excel(input_excel_file)
    df['NER_Tag'] = df['NER_Tag'].astype(str).str.strip()

    # 2. Extract total token count and per-class frequencies
    total_tokens = len(df)
    class_counts = df['NER_Tag'].value_counts()
    unique_classes = sorted(df['NER_Tag'].unique())
    num_classes = len(unique_classes)

    # 3. Calculate inverse frequency class weights formula: Total / (Num_Classes * Class_Count)
    weights_dict = {}
    print("📊 --- Calculated Loss-Penalties (Class Weights) ---")
    for cls in unique_classes:
        count = class_counts[cls]
        # Mathematical inverse balance factor
        weight = total_tokens / (num_classes * count)
        weights_dict[cls] = round(weight, 4)
        print(f"Label: [{cls:10}] | Frequency Count: {count:5} | Loss Weight Multiplier: {weight:.4f}")

    # 4. Format as a clean, ready-to-use PyTorch/TensorFlow Python code string
    code_string = "import torch\n\n"
    code_string += "# Copy-paste this ordered class list directly into your Dataset Class:\n"
    code_string += f"NER_CLASSES = {unique_classes}\n\n"
    code_string += "# Copy-paste this tensor directly into your CrossEntropyLoss(weight=NER_WEIGHTS_TENSOR):\n"

    ordered_weights = [weights_dict[cls] for cls in unique_classes]
    code_string += f"NER_WEIGHTS_TENSOR = torch.tensor({ordered_weights}, dtype=torch.float)\n"

    # 5. Write the script matrix block to a text file
    with open(output_txt_file, 'w', encoding='utf-8') as f:
        f.write(code_string)

    print("\n" + "="*60)
    print(f"✅ Success! Your Deep Learning training parameters are computed.")
    print("="*60)

    # 6. Generate an instant browser download link for the text file
    with open(output_txt_file, 'rb') as f:
        file_bytes = f.read()

    b64_encoded = base64.b64encode(file_bytes).decode()
    download_href = (
        f'<a href="data:text/plain;base64,{b64_encoded}" '
        f'download="{output_txt_file}" style="font-weight: bold; color: #0288D1; font-size: 16px;">'
        f'👉 Click Here to Download "{output_txt_file}" for Your Model Code 👈</a>'
    )

    return HTML(download_href)

# --- Runtime Trigger Setup ---
source_workbook = "0MLT__Data.xlsx"
target_code_file = "ner_model_weights.txt"

# Execute calculation and display the download text link
calculate_ner_weights_and_download(source_workbook, target_code_file)
```
